Entropy maximisation introduces a degree of randomness in the policy.

The optimal policy is defined for SAC as :

$\pi^* = \arg\max_\pi \mathbb{E} \left[\sum_{k=0}^\infty \gamma^k (R_t + \alpha \mathcal{H}) \right]$

$V_\pi(s) = \mathbb{E} \left[\sum_{k=0}^\infty \gamma^k (R_t + \alpha \mathcal{H}) \right | s_0 = s]$

The SAC concourently learns a policy $\pi_\theta$ and two Q-functions $Q_{\phi 1}$ and $Q_{\phi 2}$

Deep Deterministic Poliucy Gradient (DDPG) > TD3 > SAC

DDPG uses off-policy data and the Bellman equation to learn the Q function.

What makes DDPG off-policy ?
* The **Actor**, a deterministic policy, output the action the agent should take in state s
* The **Critic**, a Q-function, estimates the expected return from taking action a in state s, following the actor's policy.

The behavior policy is used to explore the environment and collect experience, typically by adding exploration noise to the deterministic actor's actions. The deterministic target policy is the policy being learned and improved using this collected experience.

The behavior policy is derived from the deterministic actor, rather than being an independently trained policy. For example by introducing exploration noise, entropy bonus etc.
