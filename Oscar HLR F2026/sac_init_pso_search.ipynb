Feiyang's current implementation of SAC-initialised PSO search

1. Load a pretrained SAC agent (through `.pth` model for ex) (SAC is pretrained offline)
2. Perform SAC rollout to generate beamforming candidates
3. Select the best SAC candidate (does that mean SAC is run multiple times?, or best SI checkpoints are saved over the execution of a single SAC?)
4. Run local PSO refinement on smaller search space (centered at the SAC's best solution)

Notes :

The SAC agent must be set to deterministic actions (i.e not sample from its Gaussian distribution)

if SAC returns $\mathbf x_{\mathrm{SAC}}$, the local search space becomes:

$$
\boxed{
x_{i,\min}^{\mathrm{local}}
=\max(x_{i,\min},x_{i,\mathrm{SAC}}-\Delta_i)
}
$$

$$
\boxed{
x_{i,\max}^{\mathrm{local}}
=\min(x_{i,\max},x_{i,\mathrm{SAC}}+\Delta_i)
}
$$
