$$Q^\pi(s,a)=\mathbb{E}_\pi[G_t\mid S_t=s,\ A_t=a]$$

The action-value function, or Q-function, answers the question:
"If I am in state \(s\), take action \(a\), and then continue following my policy, how much total future reward should I expect?"

In practice, the Q-function is the return accumulated from time $t=t_0$ to $t=T$ (terminal state). So it is computed over a full episode.

## 1D random walk exercise to estimate the Q-function and state-value function

The discounted return is

$$G_t \doteq \sum_{k=0}^{T-t-1}\gamma^k R_{t+k+1}$$

or equivalently,

$$G_t = R_{t+1} + \gamma R_{t+2} + \gamma^2 R_{t+3} + \cdots$$

We can rewrite this as

$$
G_t
=
R_{t+1}
+
\gamma
\left(
R_{t+2}
+
\gamma R_{t+3}
+
\cdots
\right)
$$

Therefore,

$$
G_t = R_{t+1} + \gamma G_{t+1}
$$

So, after an episode is complete, we can compute the returns efficiently by going backward using

$$
G \leftarrow R_{t+1} + \gamma G
$$

for

$$
t = T-1,\ T-2,\ \ldots,\ 0
$$

## Updating state-value function

$$Q(s,a)\leftarrow Q(s,a)+\frac{1}{N(s,a)}\left(G-Q(s,a)\right)$$


```python
import numpy as np

STATES = range(6)
ITERATIONS = 1000
GAMMA = 0.9

Q = {(state, action): 0.0 for state in STATES for action in [-1, 1]}
N = {(state, action): 0 for state in STATES for action in [-1, 1]}


def get_reward(state):
    if state == 5:
        return 1
    return 0


def take_action(state, action):
    return state + action


def execute_policy():
    return np.random.choice([-1, 1], p=[0.5, 0.5])


def generate_episode(init_state):
    episode = []
    state = init_state

    while state != 0 and state != 5:

        action = execute_policy()

        next_state = take_action(state, action)

        reward = get_reward(next_state)

        episode.append((state, action, reward))

        state = next_state

    return episode


def update_Q(episode):
    G = 0
    gamma = 0.95

    # Work backward to compute G
    for state, action, reward in reversed(episode):
        G = reward + gamma * G

        N[(state, action)] += 1

        Q[(state, action)] += (1 / N[(state, action)]) * (G - Q[(state, action)])

for i in range(100):
    n = np.random.choice([0,1,2,3,4,5])
    update_Q(generate_episode(n))

print(Q)
```
