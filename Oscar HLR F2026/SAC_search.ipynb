SAC works as follows:

1. Sample a transition from the replay buffer:
   $$
   (s_t,a_t,r_t,s_{t+1},d_t)\sim\mathcal{D}
   $$

2. Sample a next action from the current policy:
   $$
   a_{t+1}\sim\pi(\cdot|s_{t+1})
   $$

3. Evaluate the next state-action pair using the critic:
   $$
   Q(s_{t+1},a_{t+1})
   $$

4. Build the soft TD target:
   $$
   y_t
   =
   r_t
   +
   \gamma(1-d_t)
   \left[
   Q(s_{t+1},a_{t+1})
   -
   \alpha\log\pi(a_{t+1}|s_{t+1})
   \right]
   $$

5. Update the critic so that:
   $$
   Q(s_t,a_t)\approx y_t
   $$
